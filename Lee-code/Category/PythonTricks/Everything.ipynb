```python
import random

def rand5():
    return random.randint(1, 5)

def rand7():
    while True:
        x = 5 * (rand5() - 1) + rand5()   # 均匀分布在 1~25
        if x <= 21:                        # 拒绝 22~25
            return (x - 1) % 7 + 1         # 1~21 → 1~7，每个值对应 3 个 x
```
