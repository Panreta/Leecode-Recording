# Defaultdict

defaultdict(int): Create a dict with the value to be int(Can be trained to list table or something else.).

```python
from collections import defaultdict

count = defaultdict(int)
for ch in "mississippi":
    count[ch] += 1   # no KeyError even on first sight of a char
print(count)  # {'m': 1, 'i': 4, 's': 4, 'p': 2}
```

```python
group = defaultdict(list)
group["a"].append(1)
group["a"].append(2)
group["b"].append(3)
print(group)  # {'a': [1, 2], 'b': [3]}

print(f"dsf")
```
