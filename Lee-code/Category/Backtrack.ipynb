# 17. Letter Combinations of a Phone Number

```python
def letterCombinations(self, digits):
        """
        :type digits: str
        :rtype: List[str]
        """
        res = []
        digitToChar = {
            "2": "abc",
            "3": "def",
            "4": "ghi",
            "5": "jkl",
            "6": "mno",
            "7": "qprs",
            "8": "tuv",
            "9": "wxyz",
        }
        def backtrack(i,curPath):
            if len(curPath) == len(digits):# determine what to extract
                res.append(curPath)
                return

            for char in digitToChar[digits[i]]:# growing tree logic
                backtrack(i + 1, curPath + char)


        backtrack(0,"")
        return res
```

```python
def combine(n, k):
        result = []
        
        def backtrack(start, current):
            if len(current) == k:
                result.append(current[:])
                return
            
            for i in range(start, n + 1):
                current.append(i)# since we can't do backtrack(i + 1, current.append(i)), we can only add first and pop later
                backtrack(i + 1, current)
                current.pop()
        
        backtrack(1, [])
        return result
```

```python
def permuteUnique(nums):# [1,1,2], or [1_1,1_2,2] to represent the position.
    res = []
    used = [False] * len(nums) # since the elements can be duplicated, we use "used" to determine if the element is used or not
    nums.sort()

    def backtrack(path):
        if len(path) == len(nums):
            res.append(path[:])
            return
        for i in range(len(nums)):
            if used[i]:# don't use the same element in a position twice
                continue
            if i > 0 and nums[i] == nums[i-1] and not used[i-1]:# result like [1_1,1_2,2] and [1_2,1_1,2] will never happen
                continue
            used[i] = True
            path.append(nums[i])
            backtrack(path)
            path.pop()
            used[i] = False

    backtrack([])
    return res
```

78. Subsets

```python
def subsets(nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        result = []

        def backtrack(i,path):
            result.append(path[:])

            for j in range(i + 1,len(nums)):
                path.append(nums[j])
                backtrack(j,path)
                path.pop()


        backtrack(-1,[])
        return result

print(subsets([1,2,3])) 
```

52. N-Queens II

```python
def totalNQueens(self, n):
        """
        :type n: int
        :rtype: int
        """
        count = [0]
        cols = set()
        pos_diag = set()
        neg_diag = set()

        def backtrack(row):
            if row == n:
                count[0] += 1
                return

            for col in range(n):
                if col in cols or (row + col) in pos_diag or (row - col) in neg_diag:
                    continue

                cols.add(col)
                pos_diag.add(row+col)
                neg_diag.add(row - col)

                backtrack(row + 1)

                cols.remove(col)
                pos_diag.remove(row+col)
                neg_diag.remove(row - col)

        backtrack(0)
        return count[0]
```

22. Generate Parentheses

You can find the invariable inequality is closedN <= openN <= N

```python
def generateParenthesis(n):
        """
        :type n: int
        :rtype: List[str]
        """
        stack = []
        result = []

        def backtrack(openN,closedN):

            if openN == closedN == n:
                result.append("".join(stack))
                return

            if openN < n:
                stack.append("(")
                backtrack(openN + 1,closedN)
                stack.pop()
            if closedN < openN:
                stack.append(")")
                backtrack(openN,closedN + 1)
                stack.pop()

        backtrack(0,0)
        return result
```
