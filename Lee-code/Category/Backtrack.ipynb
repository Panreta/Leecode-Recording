# 698. Partition to K Equal Sum Subsets

In another word, can be divided to sum(nums)/k each, so it must be an integer

```python
def canPartitionKSubsets(self, nums: List[int], k: int) -> bool:
        if sum(nums) % k != 0:
            return False

        nums.sort(reverse=True)
        target = sum(nums) // k
        used = [False] * len(nums)

        def backtrack(i, k, subsetSum):#i: start index, k:how many left to fill, subsetSum: current sum of the subset
            if k == 0:
                return True
            if subsetSum == target:
                return backtrack(0, k - 1, 0)
            for j in range(i, len(nums)):
                if used[j] or subsetSum + nums[j] > target:
                    continue
                used[j] = True
                if backtrack(j + 1, k, subsetSum + nums[j]):
                    return True
                used[j] = False
            return False

        return backtrack(0, k, 0)
```

```python
def canPartitionKSubsets(self, nums: List[int], k: int) -> bool:
        total = sum(nums)
        if total % k != 0:
            return False

        nums.sort(reverse=True)
        target = total // k
        n = len(nums)

        def backtrack(i, k, subsetSum, mask):
            if k == 0:
                return True
            if subsetSum == target:
                return backtrack(0, k - 1, 0, mask)
            for j in range(i, n):
                if (mask & (1 << j)) == 0 or subsetSum + nums[j] > target:#j-th entry or mask is 0; means used[j] == True
                    continue
                if backtrack(j + 1, k, subsetSum + nums[j], mask ^ (1 << j)):# mask ^ (1 << j): flip
                    return True
                if subsetSum == 0:# for the case like [4,4,4,4,4,4], k = 4; it will work
                    return False
            return False

        return backtrack(0, k, 0, (1 << n) - 1)# [1] * len(nums),1 is false,0 is true
```

```python
def makesquare(self, matchsticks):
        """
        :type matchsticks: List[int]
        :rtype: bool
        """
        nums = matchsticks
        k = 4
        if sum(nums) % k != 0:
            return False

        nums.sort(reverse=True)
        target = sum(nums) // k
        used = [False] * len(nums)

        def backtrack(i, k, subsetSum):#i: start index, k:how many left to fill, subsetSum: current sum of the subset
            if k == 0:
                return True
            if subsetSum == target:
                return backtrack(0, k - 1, 0)
            for j in range(i, len(nums)):
                if used[j] or subsetSum + nums[j] > target:
                    continue
                used[j] = True
                if backtrack(j + 1, k, subsetSum + nums[j]):
                    return True
                used[j] = False
                if subsetSum == 0:
                    return False
            return False

        return backtrack(0, k, 0)
```

# 2850. Minimum Moves to Spread Stones Over Grid

```python
def minimumMoves(grid):
        """
        :type grid: List[List[int]]
        :rtype: int
        """
        more = []
        less = []
        for i in range(3):
            for j in range(3):
                if grid[i][j] > 1:
                    more.extend([(i, j)] * (grid[i][j] - 1))
                elif grid[i][j] == 0:
                    less.append((i, j))

        used = [False] * len(less)
        best = float('inf')

# intuition, move from more to less
        def backtrack(idx, cost):
            nonlocal best
            if cost >= best:
                return
            if idx == len(more):
                best = min(best, cost)
                return
            for k in range(len(less)):
                if used[k]:
                    continue
                used[k] = True
                dist = abs(more[idx][0]-less[k][0]) + abs(more[idx][1]-less[k][1])# how many step should I move
                backtrack(idx + 1, cost + dist)
                used[k] = False

        backtrack(0, 0)
        return best
```

# 1723. Find Minimum Time to Finish All Jobs

```python

def minimumTimeRequired(jobs: List[int], k: int) -> int:
    jobs.sort(reverse=True)           # big balls first -> early pruning
    loads = [0] * k                   # current load of each box (worker)
    best = sum(jobs)                  # valid upper bound: one worker does everything

    def backtrack(i: int, cur_max: int) -> None:
        nonlocal best
        if cur_max >= best:           # this branch can't beat best
            return
        if i == len(jobs):            # all balls placed
            best = cur_max
            return
        seen = set()                  # loads already tried for this ball
        for w in range(k):
            if loads[w] in seen:      # the allocatiion we talked about before
                continue
            if loads[w] + jobs[i] >= best:
                continue              # bound pruning
            seen.add(loads[w])
            loads[w] += jobs[i]       
            backtrack(i + 1, max(cur_max, loads[w]))
            loads[w] -= jobs[i]       

    backtrack(0, 0)
    return best
```

# Board Problem

For this part, we gonna talk about how to solve the classic board games such as sudoku.

```python
def solveSudoku(board):
        """
        :type board: List[List[str]]
        :rtype: None Do not return anything, modify board in-place instead.
        """
        rows = [set() for _ in range(9)]
        cols = [set() for _ in range(9)]
        boxes = [set() for _ in range(9)]
        empty = []

        def box_mapping(i,j):
            return i // 3 * 3 + j // 3

        def backtrack(idx):
            if idx == len(empty):
                return True
            i,j = empty[idx]
            b = box_mapping(i,j)
            for d in "123456789":
                if d in rows[i] or d in cols[j] or d in boxes[b]:
                    continue
                board[i][j] = d
                rows[i].add(d),cols[j].add(d),boxes[b].add(d)
                if backtrack(idx + 1):
                    return True
                board[i][j] = "."

                rows[i].remove(d),cols[j].remove(d),boxes[b].remove(d)
            return False
        for i in range(9):
            for j in range(9):
                ele = board[i][j]
                if ele.isdigit():
                    rows[i].add(ele)
                    cols[j].add(ele)
                    boxes[box_mapping(i,j)].add(ele)
                else:
                    empty.append((i,j))
        backtrack(0)
```

```python
def solveNQueens(n: int) -> List[List[str]]:
    res = []
    board = [["."] * n for i in range(n)]

    def backtrack(r):
        if r == n:
            copy = ["".join(row) for row in board]
            res.append(copy)
            return
        for c in range(n):
            if isSafe(r, c, board):
                board[r][c] = "Q"
                backtrack(r + 1)
                board[r][c] = "."

    backtrack(0)
    return res

def isSafe(r: int, c: int, board):# we start from a lower row, so only check upper lefft, upper right
    row = r - 1
    while row >= 0:
        if board[row][c] == "Q":
            return False
        row -= 1

    row, col = r - 1, c - 1
    while row >= 0 and col >= 0:
        if board[row][col] == "Q":
            return False
        row -= 1
        col -= 1

    row, col = r - 1, c + 1
    while row >= 0 and col < len(board):
        if board[row][col] == "Q":
            return False
        row -= 1
        col += 1
    return True
```
