74. Search a 2D Matrix

```python
def searchMatrix(matrix, target):
        """
        :type matrix: List[List[int]]
        :type target: int
        :rtype: bool
        """
        m,n = len(matrix) , len(matrix[0])
        left, right = 0,m - 1
        while left <= right:
            mid = (left + right)//2
            if matrix[mid][0] == target:
                return True
            elif matrix[mid][0] > target:
                right = mid - 1
            else:
                left = mid + 1
        ## now right < left
        row = right         
        if row < 0:          # target smaller than everything
            return False
        
        left, right = 0,n - 1
        while left <= right:## <=
            mid = (left + right)//2
            if matrix[row][mid] == target:
                return True
            elif matrix[row][mid] > target:
                right = mid - 1
            else:
                left = mid +  1
        return False
```

162. Find Peak Element

```python
def findPeakElement(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        m = len(nums)
        left, right = 0,m - 1
        while left <= right:
            mid = (left + right)//2
            if mid > 0 and nums[mid - 1] > nums[mid]:
                right = mid - 1
            elif mid < m - 1 and nums[mid + 1] > nums[mid]:
                left = mid + 1
            else:
                return mid
```

33. Search in Rotated Sorted Array

```python
def search(nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right)//2
            if nums[mid] == target:
                return mid

            if nums[left] <= nums[mid]:      # left half is sorted
                if nums[left] <= target < nums[mid]:   # only when gurenttee we get the totally sorted range, can we use bs
                    right = mid - 1
                else:
                    left = mid + 1
            elif nums[mid] <= nums[right]:   # right half is sorted
                if nums[mid] < target <= nums[right]:  # target in sorted right range
                    left = mid + 1
                else:
                    right = mid - 1
        return -1
```

34. Find First and Last Position of Element in Sorted Array

```python
from collections import bisect
def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        left = bisect.bisect_left(nums,target)
        if left >= len(nums) or nums[left] != target:
            return [-1,-1]

        right = bisect.bisect_right(nums,target) - 1
        return [left,right]
```

```python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        left = self.binarySearch(nums, target,leftBias = True)
        right = self.binarySearch(nums, target, leftBias = False)
        return [left,right]

    def binarySearch(self,nums,target,leftBias):
        left,right = 0, len(nums) - 1
        i = -1 # if left already > right, remember [-1,-1]
        while left <= right:
            mid = (left + right) // 2
            if target > nums[mid]:
                left = mid + 1
            elif target < nums[mid]:
                right = mid - 1
            else:
                i = mid
                if leftBias:
                    right = mid - 1
                else:
                    left = mid + 1

        return i
```

4. Median of Two Sorted Arrays

```python
class Solution:
    def findMedianSortedArrays(self, nums1, nums2):
        total = len(nums1) + len(nums2)
        left_k  = (total + 1) // 2 # if total is odd, it is the median; if even, it is the left median
        right_k = (total + 2) // 2 # if odd, it is the median; if even, it is the right median.
        return (self.kth(nums1, nums2, left_k) + self.kth(nums1, nums2, right_k)) / 2.0 # if odd, the equation can be simplified, find the median in the two lists
    # not only integer, so use /2.0 to get float

    def kth(self, a, b, k):
        # always keep a as the shorter array
        if len(a) > len(b):
            return self.kth(b, a, k)

        # base cases
        if len(a) == 0:# maintain two lists must have elements
            return b[k - 1]
        if k == 1: # simple case
            return min(a[0], b[0])

        elim_a = min(len(a), k // 2) # elim_a <= len(a), so you can do idx; elim_a <= k//2, so no k-th element contains in 0,...,elim_a
        elim_b = min(len(b), k // 2)

        if a[elim_a - 1] > b[elim_b - 1]:# discard the smaller one of b for it def has no k-th element
            return self.kth(a, b[elim_b:], k - elim_b)# since I discarded elim_b elements from b, the k-th element is now (k - elim_b)th element
        else:
            return self.kth(a[elim_a:], b, k - elim_a)
```
