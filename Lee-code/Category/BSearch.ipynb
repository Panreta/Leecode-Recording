```python
def searchInsert(nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        left, right = 0, len(nums) - 1

        while left <= right:## = when nums[mid] == target may work, this time left,right -> correct index
            mid = (left + right)//2
            if nums[mid] == target:
                return mid
            elif nums[mid] > target:
                right = mid - 1
            else:
                left = mid +  1

        return left ## may can't find
```

74. Search a 2D Matrix

```python
def searchMatrix(matrix, target):
        """
        :type matrix: List[List[int]]
        :type target: int
        :rtype: bool
        """
        m,n = len(matrix) , len(matrix[0])
        left, right = 0,m - 1
        while left <= right:
            mid = (left + right)//2
            if matrix[mid][0] == target:
                return True
            elif matrix[mid][0] > target:
                right = mid - 1
            else:
                left = mid + 1
        ## now right < left
        row = right         
        if row < 0:          # target smaller than everything
            return False
        
        left, right = 0,n - 1
        while left <= right:## <=
            mid = (left + right)//2
            if matrix[row][mid] == target:
                return True
            elif matrix[row][mid] > target:
                right = mid - 1
            else:
                left = mid +  1
        return False
```

162. Find Peak Element

```python
def findPeakElement(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        m = len(nums)
        left, right = 0,m - 1
        while left <= right:
            mid = (left + right)//2
            if mid > 0 and nums[mid - 1] > nums[mid]:
                right = mid - 1
            elif mid < m - 1 and nums[mid + 1] > nums[mid]:
                left = mid + 1
            else:
                return mid
```

33. Search in Rotated Sorted Array

```python
def search(nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right)//2
            if nums[mid] == target:
                return mid

            if nums[left] <= nums[mid]:      # left half is sorted
                if nums[left] <= target < nums[mid]:   # only when gurenttee we get the totally sorted range, can we use bs
                    right = mid - 1
                else:
                    left = mid + 1
            elif nums[mid] <= nums[right]:   # right half is sorted
                if nums[mid] < target <= nums[right]:  # target in sorted right range
                    left = mid + 1
                else:
                    right = mid - 1
        return -1
```
