9. Palindrome Number

```python
def isPalindrome(self, x):
        if x < 0:
            return False
      
        x = str(x)
        n = len(x)
        i = n // 2 - 1
        if n % 2 == 0:
            
            j = n // 2
            while i >= 0 and j <= n - 1:
                if x[i] == x[j]:
                    i -= 1
                    j += 1
                else:
                    return False
            return True
        
        else: 
            i += 1
            for j in range(1,n//2 + 1):
                if x[i - j] != x[i + j]:
                    return False
            return True
```

66. Plus One

```python
def plusOne(self, digits):
        """
        :type digits: List[int]
        :rtype: List[int]
        """
        n = len(digits)
        carry = 1

        for i in range(n - 1,-1,-1):
            value = digits[i] + carry
            digits[i] = value % 10
            carry = value // 10
        if carry == 1:
            digits = [1] + digits

        return digits
```

172. Factorial Trailing Zeroes

```python
def trailingZeroes(self, n):
        """
        :type n: int
        :rtype: int
        """
        ans = 0

        while n:
            n //= 5
            ans += n
        return ans
```

```python
def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        if x < 2:
            return x

        l,r = 0, x // 2
        while l <= r:
            mid = (l + r) // 2
            if mid * mid == x:
                return mid
            elif mid * mid > x:
                r = mid - 1
            else:
                l = mid + 1
        return r
```

50. Pow(x, n)

```python
def myPow(x, n):
        """
        :type x: float
        :type n: int
        :rtype: float
        """
        res = x
        count = 1 
        shortCut = n
        while shortCut != 1:
            count *= 2
            shortCut //= 2
            res *= res
            print(f"res is {res},n is {n},count is {count}")
        resdue = n - count
        print(f"resdue is {resdue}")
        for _ in range(resdue):
            print(f"res is {res}")
            res *= x
        
        return res
print(myPow(2,10))
```
