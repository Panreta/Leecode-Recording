67. Add Binary

```python
def get_digit_and_advance(s, idx):
    if idx >= 0:
        return int(s[idx]), idx - 1
    return 0, idx

def addBinary(a, b):
    i,j = len(a) - 1, len(b) - 1
    carry = 0
    res = ""
    while i >= 0 or j >= 0 or carry:
        da, i = get_digit_and_advance(a, i)
        db, j = get_digit_and_advance(b, j)
        total = carry + da + db
        res += str(total % 2) # is there any 1 left
        carry = total // 2# is it larger than 2 to get carry

    return res[::-1]


```

190. Reverse Bits

n =$\sum_{i=0}^n a_i*2^i$ , wanna get $\sum_{i=0}^n a_{n - i}*2^i=a_0 * 2^n +...+a_n$

n % 2 = $a_0$,
i = 0, ans = 0 + $a_0$,n = $\sum_{i=1}^n a_i*2^i$
i = 1, ans = $a_0$ * 2 + $a_1$, n = $\sum_{i=2}^n a_i*2^i$...

```python
def reverseBits(n):
    ans = 0

    for i in range(32):
        ans = ans * 2 + (n % 2)
        n = n // 2

    return ans
```

Translate to binary code

```python
def reverseBits(n):
    result = 0

    for i in range(32):
        bit = n & 1# a_2
        result += (result << 1) | bit #(a_0,a_1,__) - >(a_0,a_1,a_2)
        n >> 1 #(a_n,...,a_3)
    return result
```

191. Number of 1 Bits

```python
def hammingWeight(n):
        """
        :type n: int
        :rtype: int
        """
        br = str(format(n,'b'))
        result = 0
        for i in range(len(br)):
            if br[i] == "1":
                result += 1
        return result
```

```python
def hammingWeight(n):
        """
        :type n: int
        :rtype: int
        """
        count = 0
        while n != 0:
            count += n & 1
            n >>= 1
        return count
```

```python
def singleNumber(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        index = 0
        for num in nums: # a ^ b ^ a == a^a^b = b, which one left is the single number
            index ^= num
        return index
```
