15. 3Sum, find 3 elements to sum to 0.

Solution, fix nums[i] < 0, use l,r search in [i+1:] about the answer.

```python
def threeSum(nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        nums.sort()
        n = len(nums)
        ans = []
        for i in range(n - 2):
            if nums[i] > 0: # At this step, all the element will be larger than 0, a_i > 0 \sum a_i > 0.
                break
            if i > 0 and nums[i - 1] == nums[i]: # like [-1 -1 -1 2], it prevents to get two [-1 -1 2], a doesn't have to be the same value
                continue 

            l, r = i + 1, n - 1
            while l < r:
                if nums[i] + nums[l] + nums[r] < 0:
                    l += 1
                elif nums[i] + nums[l] + nums[r] > 0:
                    r -= 1
                else:
                    ans.append([nums[i], nums[l], nums[r]])
                    l += 1
                    r -= 1
                     # what if we have [-4,...,0,0,...,4,4],when [-4,0,4] selected next -4 0 can be
                    # selected only if nums[l] and nums[r] don't change,just change l is ok
                    while l < r and nums[l - 1] == nums[l]: # before we only prevent i, here is also for l and r.
                        l += 1
       
        return ans
```

```python
def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        n = len(s)
        l,r = 0,n - 1
        s = s.lower()

        while l < r:
            if not s[l].isalnum():
                l += 1
                if l < n:
                    continue
            
             
            if not s[r].isalnum():
                r -= 1
                if r >= 0:
                    continue
             
            if s[l] != s[r]:
                return False
            l += 1
            r -= 1
        return True
```

88. Merge Sorted Array

Do from the back: Eg: num1 = [1,2,3,0,0,0] nums2 = [2,5,6]; for you don't have to move the 0 back

```python
def merge(nums1, m, nums2, n):
        """
        :type nums1: List[int]
        :type m: int
        :type nums2: List[int]
        :type n: int
        :rtype: None Do not return anything, modify nums1 in-place instead.
        """
        i,j,final_in_nums1 = m - 1,n - 1, m+n - 1

        while i >= 0 and j >= 0:
            if nums1[i] >= nums2[j]:
                nums1[final_in_nums1] = nums1[i]
                i -= 1
            else:
                nums1[final_in_nums1] = nums2[j]
                j -= 1

            final_in_nums1 -= 1

        while j >= 0:
            nums1[final_in_nums1] = nums2[j]
            final_in_nums1 -= 1
            j -= 1 
```
