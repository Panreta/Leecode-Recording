# 200. Number of Islands

```python
def numIslands(grid):
        """
        :type grid: List[List[str]]
        :rtype: int
        """
        m,n = len(grid),len(grid[0])
        count = 0

        def dfs(i,j): # recursive equation + halting condition
            if ((i < 0 or i >= m) or
                (j < 0 or j >= n) or
                (grid[i][j] == '0')):
                return
            grid[i][j] = '0'
            dfs(i - 1,j)
            dfs(i + 1,j)
            dfs(i,j - 1)
            dfs(i,j + 1)

        # iteration
        for i in range(m):
            for j in range(n):
                if grid[i][j] == '1':
                    dfs(i,j)
                    count += 1
        return count
                    
```

```python
from collections import deque

def numIslands(grid):
    """
    :type grid: List[List[str]]
    :rtype: int
    """
    m, n = len(grid), len(grid[0])
    count = 0

    def bfs(i, j):
        q = deque([(i, j)])
        grid[i][j] = '0'
        while q:
            ci, cj = q.popleft()
            for di, dj in [(-1,0),(1,0),(0,-1),(0,1)]:
                ni, nj = ci + di, cj + dj
                if 0 <= ni < m and 0 <= nj < n and grid[ni][nj] == '1':
                    grid[ni][nj] = '0'
                    q.append((ni, nj))

    for i in range(m):
        for j in range(n):
            if grid[i][j] == '1':
                bfs(i, j)
                count += 1
    return count
```

# 261. Graph Valid Tree

```python
def validTree(n, edges):
        if not edges:
            return n == 1
        seen = set()
        seen.add(edges[0][0])
        seen.add(edges[0][1])
        remaining = edges[1:]
        remaining.sort(key=lambda x: x[0])

        changed = True
        while changed and remaining:
            changed = False
            next_remaining = []
            for a, b in remaining:
                if a in seen and b in seen:
                    return False
                elif a in seen or b in seen:
                    seen.add(a)
                    seen.add(b)
                    changed = True
                else:
                    next_remaining.append((a, b))
            remaining = next_remaining

        return len(seen) == n
```

# DFS

```python
def validTree(n, edges):
    if len(edges) > (n - 1):
        return False

    adj = [[] for _ in range(n)]
    for u, v in edges:
        adj[u].append(v)
        adj[v].append(u)

    visit = set()
    def dfs(node, par):
        if node in visit:
            return False

        visit.add(node)
        for nei in adj[node]:
            if nei == par:# If we are testing the adjacent node is exactly the par(testing dfs(1,0) apparently 0 == node should be continue)
                continue
            if not dfs(nei, node):
                return False
        return True

    return dfs(0, -1) and len(visit) == n# dfs no mistake and tested all the nodes; -1 random start compare point
```

# Disjoint Set Union

```python
class DSU:
    def __init__(self, n):
        self.comps = n
        self.Parent = list(range(n + 1))#[0,...,n]
        self.Size = [1] * (n + 1)# nodes in the sub tree

    def find(self, node):
        if self.Parent[node] != node:
            self.Parent[node] = self.find(self.Parent[node])
        print(f"in find, node is {node},return {self.Parent[node]}")
        return self.Parent[node]

    def union(self, u, v):
        pu = self.find(u)
        pv = self.find(v)
      
        if pu == pv:# where to show the loop
            return False

        self.comps -= 1
        if self.Size[pu] < self.Size[pv]:
            pu, pv = pv, pu
            
        self.Size[pu] += self.Size[pv]
        self.Parent[pv] = pu
        return True

    def components(self):
        return self.comps

def validTree(n: int, edges):
    if len(edges) > n - 1:# |V|=n,|E|=n - 1
        return False

    dsu = DSU(n)
    for u, v in edges:
        if not dsu.union(u, v):
            return False
    return dsu.components() == 1

n = 5
edges = [[0,1],[0,2],[0,3],[1,4]]
validTree(n,edges)
```

# 130. Surrounded Regions

```python
def solve(board):
    """
    :type board: List[List[str]]
    :rtype: None Do not return anything, modify board in-place instead.
    """
    if not board or not board[0]:
        return
    m, n = len(board), len(board[0])

    def dfs(i, j, mark):
        if i < 0 or i >= m or j < 0 or j >= n or board[i][j] != 'O':
            return
        board[i][j] = mark
        dfs(i - 1, j, mark)
        dfs(i + 1, j, mark)
        dfs(i, j - 1, mark)
        dfs(i, j + 1, mark)

    for i in range(m):
        dfs(i, 0, 'P')
        dfs(i, n - 1, 'P')
    for j in range(n):
        dfs(0, j, 'P')
        dfs(m - 1, j, 'P')

    for i in range(m):
        for j in range(n):
            if board[i][j] == 'O':
                dfs(i, j, 'X')

    for i in range(m):
        for j in range(n):
            if board[i][j] == 'P':
                board[i][j] = 'O'
```

# 733. Flood Fill

```python
def floodFill(image, sr, sc, color):
        """
        :type image: List[List[int]]
        :type sr: int
        :type sc: int
        :type color: int
        :rtype: List[List[int]]
        """
        m,n = len(image),len(image[0])
        START = image[sr][sc]
        if color == START:# for if start is 0 and others are all 0, the algo will go in the region forever
            return image
        def dfs(i,j):
            if (i < 0 or i >= m or j < 0 or j >= n or image[i][j] != START):
                return
            image[i][j] = color
            dfs(i - 1,j)
            dfs(i + 1,j)
            dfs(i,j - 1)
            dfs(i,j + 1)
            return
        dfs(sr,sc)
        return image
```

# 1971. Find if Path Exists in Graph

```python
class Solution(object):
    def validPath(self, n, edges, source, destination):
        """
        :type n: int
        :type edges: List[List[int]]
        :type source: int
        :type destination: int
        :rtype: bool
        """
        adj = {i: [] for i in range(n)}
        for u, v in edges:
            adj[u].append(v)
            adj[v].append(u)

        visited = set()

        def dfs(node):
            if node == destination:
                return True
            visited.add(node)
            for neighbor in adj[node]:
                if neighbor not in visited:
                    if dfs(neighbor):
                        return True
            return False

        return dfs(source)
```
