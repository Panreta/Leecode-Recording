# 200. Number of Islands

# DFS

```python
def numIslands(grid):
        """
        :type grid: List[List[str]]
        :rtype: int
        """
        m,n = len(grid),len(grid[0])
        count = 0

        def dfs(i,j):
            if ((i < 0 or i >= m) or
                (j < 0 or j >= n) or
                (grid[i][j] == '0')):
                return
            grid[i][j] = '0'
            dfs(i - 1,j)
            dfs(i + 1,j)
            dfs(i,j - 1)
            dfs(i,j + 1)

        for i in range(m):
            for j in range(n):
                if grid[i][j] == '1':
                    dfs(i,j)
                    count += 1
        return count
                    
```

# 1971. Find if Path Exists in Graph

```python
class Solution(object):
    def validPath(self, n, edges, source, destination):
        """
        :type n: int
        :type edges: List[List[int]]
        :type source: int
        :type destination: int
        :rtype: bool
        """
        adj = {i: [] for i in range(n)}
        for u, v in edges:
            adj[u].append(v)
            adj[v].append(u)

        visited = set()

        def dfs(node):
            if node == destination:
                return True
            visited.add(node)
            for neighbor in adj[node]:
                if neighbor not in visited:
                    if dfs(neighbor):
                        return True
            return False

        return dfs(source)
```
