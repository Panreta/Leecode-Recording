123. Best Time to Buy and Sell Stock III

Hold at most 2 stock each time. Imagine every time holding the stocks is the state, and whether to sell is an action.

```python
def maxProfit(prices):
    """
    :type prices: List[int]
    :rtype: int
    """
    if not prices:
        return 0
    
    # Track 4 states: after 1st buy, 1st sell, 2nd buy, 2nd sell
    buy1 = float('-inf')# if you don't buy, the penalty is large
    sell1 = 0 # sell >=0, since comparing max, just set 0.
    buy2 = float('-inf')
    sell2 = 0
    
    for price in prices:
        # First transaction, every step, just compare if we need to make a move
        buy1 = max(buy1, -price)           # After the 1st buy, -inf < buy1 <= 0
        sell1 = max(sell1, buy1 + price)   # Sell after first buy, price_now - (-buy1) = price_now - price_prev
        
        # Second transaction
        buy2 = max(buy2, sell1 - price)    
        sell2 = max(sell2, buy2 + price)   
    
    return sell2
```

```python
# At most k transactions
def maxProfit(prices, k):
    buy  = [float('-inf')] * k
    sell = [0] * k

    for price in prices:
        for i in range(k):
            buy[i]  = max(buy[i],  (sell[i-1] if i > 0 else 0) - price)# if i ==0, just -price as problem 3; 
            # i > 0, see if we can get benefit from the prev transaction
            sell[i] = max(sell[i], buy[i] + price)

    return sell[-1]
```

```python
def canJump(nums):
        """
        :type nums: List[int]
        :rtype: bool
        """
        fastest = 0

        for i in range(len(nums)):
            if i > fastest:
                return False
            fastest = max(fastest, i + nums[i])

        return True
        
print(canJump([2,3,1,1,4]))
```

```python
def jump(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        jump, fastest, end = 0,0,0
        for i in range(len(nums) - 1):# you don't have to jump at late
            fastest = max(fastest,i + nums[i])
            if i == end: # it claims the jump truly happens
                jump += 1
                end = fastest

        return jump
```

```python
def canCompleteCircuit(gas, cost):
        """
        :type gas: List[int]
        :type cost: List[int]
        :rtype: int
        """
        if sum(gas) < sum(cost):
            return -1
        
        # exist one station that if we start from it, we can complete the circuit
        # Otherwise, \sum_{i=0}^{n-1} (gas[i] ) < \sum_{i=0}^{n-1} (cost[i]), i is the arbitrary starting station

        remain = 0
        start = 0
        for i in range(len(gas)):
            remain += gas[i] - cost[i]
            if remain < 0:
                start = i + 1
                remain = 0
        return start
```

135. Candy

```python
def candy(ratings):
        """
        :type ratings: List[int]
        :rtype: int
        """
        n = len(ratings)
        if n == 1:
            return 1

        if n == 2:
            if ratings[1] - ratings[0] != 0:
                return 3

        # n > 2
        candys = [1] * n
        for i in range(1,n):
            if ratings[i] > ratings[i - 1]:# check higher than the left neighbor
                candys[i] = candys[i - 1] + 1
            
        for i in range(n - 2, -1, -1):
            if ratings[i] > ratings[i + 1]:# check higher than the right neighbor
                candys[i] = max(candys[i + 1] + 1,candys[i])

        return sum(candys)
```

42. Trapping Rain Water

```python
def trap(height):
        n = len(height)
        if n == 0:
            return 0

        leftMax = [0] * n # leftMax[i]: before i, the maximum height
        rightMax = [0] * n

        leftMax[0] = height[0]
        for i in range(1, n):
            leftMax[i] = max(leftMax[i - 1], height[i])

        rightMax[n - 1] = height[n - 1]
        for i in range(n - 2, -1, -1):
            rightMax[i] = max(rightMax[i + 1], height[i])

        res = 0
        for i in range(n):
            res += min(leftMax[i], rightMax[i]) - height[i]
        return res
```

Count the palindromic in a simple way.

```python

def countPalindromicSubsequence(s: str) -> int:
    count = 0
    for c in set(s):
        first = s.index(c)
        last = s.rindex(c)
        if last - first > 1:
            middle = set(s[first + 1 : last])
            count += len(middle)
    return count
```

```python
class TrieNode:
    def __init__(self):
        self.children = [None] * 26
        self.is_end = False

class Trie(object):
    def __init__(self):
        self.root = TrieNode()

    def insert(self,word):
        curr = self.root
        for char in word:
            idx = ord(char) - ord("a")
            if curr.children[idx] ==  None:
                curr.children[idx] = TrieNode()
            curr = curr.children[idx]# move on to the children
        curr.is_end = True # end of the word, get a closure
    
    def search(self,word):
        curr = self.root
        for char in word:
            idx = ord(char) - ord('a')
            if curr.children[idx] == None:
                return False
            curr = curr.children[idx]
        return curr.is_end
    
    def startsWith(self,prefix):
        curr = self.root
        for char in prefix:
            idx = ord(char) - ord('a')
            if curr.children[idx] == None:
                return False
            curr = curr.children[idx]
        return True


def longestCommonPrefix(strs):
#         """
#         :type strs: List[str]
#         :rtype: str
#         """
    root = Trie()
    for word in strs:
        root.insert(word)
    
    curr = strs[0]
    for i in range(len(curr),-1,-1):
        if root.startsWith(curr[:i]):
            return curr[:i]
    return ""
```

```python
def sortColors(nums):
        """
        :type nums: List[int]
        :rtype: None Do not return anything, modify nums in-place instead.
        """
        k = 0
        for i in range(len(nums) - 1, -1, -1):
            
            while nums[i] == 0:
                if i <= k:
                    break
                if nums[k] != 0:
                    nums[i],nums[k] = nums[k],nums[i]
                k += 1
        if nums[k] == 0:
            k += 1
        # Ideally, after this for loop, [:k + 1] are zero.
        
        m = k
        for i in range(len(nums) - 1, k -1 , -1):
            
            while nums[i] == 1:
                if i <= m:
                    break
                if nums[m] != 1:
                    nums[i],nums[m] = nums[m],nums[i]
                m += 1
        return nums
```

304. Range Sum Query 2D - Immutable

```python
class NumMatrix(object):

    def __init__(self, matrix):
        """
        :type matrix: List[List[int]]
        """
        self.prefix = [len(matrix[0])*[0] for _ in range(len(matrix))]
        for row in range(len(matrix)):
            self.prefix[row][0] = matrix[row][0]
            for col in range(1,len(matrix[0])):
                self.prefix[row][col] = self.prefix[row][col - 1] + matrix[row][col]        

    def sumRegion(self, row1, col1, row2, col2):
        """
        :type row1: int
        :type col1: int
        :type row2: int
        :type col2: int
        :rtype: int
        """
        res = 0
        for row in range(row1, row2 + 1):
            if col1 > 0:
                res += self.prefix[row][col2] - self.prefix[row][col1 - 1]
            else:
                res += self.prefix[row][col2]
        return res
```

36. Valid Sudoku

```python
def isValidSudoku(self, board):
        """
        :type board: List[List[str]]
        :rtype: bool
        """
        rows = [set() for _ in range(9)]
        cols = [set() for _ in range(9)]
        boxs = [set() for _ in range(9)]

        for r in range(len(board)):
            for c in range(len(board[0])):
                idx = (r // 3 ) * 3 + (c // 3)
                cur = board[r][c]
                if cur == ".":
                    continue
                if cur in rows[r] or cur in cols[c] or cur in boxs[idx]:
                    return False
                rows[r].add(cur)
                cols[c].add(cur)
                boxs[idx].add(cur)
        return True
```

```python
def longestPalindrome(s: str) -> int:
    from collections import Counter
    counts = Counter(s)
    length = 0
    has_odd = False
    for c in counts.values():
        length += c - (c % 2)
        if c % 2 == 1:
            has_odd = True
    return length + 1 if has_odd else length
```

```python
def subarraySum(nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        n = len(nums)
        if n == 1 and nums[0] == k:
            return 1
        l,r = 0,0
        res = 0
        prefix = 0
        while r < n and l <= r:
            prefix += nums[r]
            r += 1
            if prefix >= k:
                if prefix == k:
                    res += 1
                prefix -= nums[l]
                l += 1
            
            
            
        return res

print(subarraySum([1,1,1],2))
```

41. First Missing Positive

```python
def firstMissingPositive(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        i = 0
        n = len(nums)

        while i < n:
            if nums[i] <= 0 or nums[i] > n:
                i += 1
                continue
            
            idx = nums[i] - 1
            if nums[i] != nums[idx]:
                nums[i], nums[idx] = nums[idx], nums[i]
            else:
                i += 1
        
        for i in range(n):
            if nums[i] != i + 1:
                return i + 1
        return n + 1
```
