215. Kth Largest Element in an Array

```python
def findKthLargest(nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        return print(heapq.nlargest(k,nums))

nums = [3, 2, 1, 5, 6, 4]
k = 2
findKthLargest(nums, 90)
```

502. IPO

```python
def findMaximizedCapital(k, w, profits, capital):
        """
        :type k: int
        :type w: int
        :type profits: List[int]
        :type capital: List[int]
        :rtype: int
        """
        maxProfit = []
        minCapital = [(c,p) for c,p in zip(capital,profits)]
        heapq.heapify(minCapital)# sort by capital

        for _ in range(k):
            while minCapital and minCapital[0][0] <= w:
                c, p = heapq.heappop(minCapital)
                heapq.heappush(maxProfit,-p)#build a min heap

            if not maxProfit:
                break

            w += -heapq.heappop(maxProfit)

        return w
```
