380. Insert Delete GetRandom O(1)

Simple version, but need to read the whole list when removing, so it's O(n) instead of O(1)

```python
class RandomizedSet(object):
    
    def __init__(self):
        self.set = []

    def insert(self, val):# not present: true; present: false
        """
        :type val: int
        :rtype: bool
        """
        if val in self.set:
            return False
        else:
            self.set.append(val)
            return True
      
    def remove(self, val):#present: true; 
        """
        :type val: int
        :rtype: bool
        """
        if val not in self.set:
            return False
        else:
            self.set.remove(val)
            return True
    

    def getRandom(self):
        """
        :rtype: int
        """
        import random
        return random.choice(self.set)
     
        
```

```python
class RandomizedSet(object):
    
    def __init__(self):
        self.dict = {}
        self.array = []

    def insert(self, val):
        """
        :type val: int
        :rtype: bool
        """
        if val in self.dict:
            return False
        self.array.append(val)
        self.dict[val] = len(self.array) - 1 # like a timestamp, also the idx in array
        #print 'added', val, ':', self.array, self.dict
        return True

    def remove(self, val):
        """
        :type val: int
        :rtype: bool
        """
        if val not in self.dict:
            return False
        pos = self.dict[val]
        
        val2 = self.array[-1]
        self.array[pos] = val2# put the last element into the position will be removed from list
        self.dict[val2] = pos # update the index of the last element
        self.array.pop()
        del self.dict[val]
        #print 'removed', val, ':', self.array
        return True
        
    def getRandom(self):
        """
        :rtype: int
        """
        from random import choice
        return choice(self.array)
```

# 208. Implement Trie (Prefix Tree)

```python
class TrieNode:
    def __init__(self):
        self.children = [None] * 26
        self.is_end = False

class Trie(object):
    def __init__(self):
        self.root = TrieNode()

    def insert(self,word):
        curr = self.root
        for char in word:
            idx = ord(char) - ord("a")
            if curr.children[idx] ==  None:
                curr.children[idx] = TrieNode()
            curr = curr.children[idx]# move on to the children
        curr.is_end = True # end of the word, get a closure
    
    def search(self,word):
        curr = self.root
        for char in word:
            idx = ord(char) - ord('a')
            if curr.children[idx] == None:
                return False
            curr = curr.children[idx]
        return curr.is_end
    
    def startsWith(self,prefix):
        curr = self.root
        for char in prefix:
            idx = ord(char) - ord('a')
            if curr.children[idx] == None:
                return False
            curr = curr.children[idx]
        return True



```
