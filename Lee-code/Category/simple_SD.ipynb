```python
class RandomizedSet(object):
    
    def __init__(self):
        self.set = []
        

    def insert(self, val):# not present: true; present: false
        """
        :type val: int
        :rtype: bool
        """
        if val in self.set:
            return False
        else:
            self.set.append(val)
            return True
    
    def remove(self, val):#present: true; 
        """
        :type val: int
        :rtype: bool
        """
        if val not in self.set:
            return False
        else:
            self.set.remove(val)
            return True

    def getRandom(self):
        """
        :rtype: int
        """
        import random
        return random.choice(self.set)
        
```
