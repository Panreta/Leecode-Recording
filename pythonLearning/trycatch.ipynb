IOError: About reading, writing, and opening the files.

```python
#!/usr/bin/python
# -*- coding: UTF-8 -*-

try:
    fh = open("testfile", "w")
    fh.write("这是一个测试文件，用于测试异常!!")
except IOError:
    print("Error: 没有找到文件或读取文件失败")
else:
    print("内容写入文件成功")
    fh.close()
```

TypeError: Wrong Type handling

```python
try:
    n = 4
    if not isinstance(n, list):
        raise TypeError
    if n[0] == []:
        print(True)
except TypeError:
    print("Wrong Type")
```

ZeroDivisionError

```python
result = 1// 0
print("Result:", result)
```

```python
try:
    result = 1// 0
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

General exception

```python
def divide(x, y):
    try:
        result = x // y
        print("Result:", result)
    except Exception as e:
        print("Error:", e)

divide(10, "2")
divide(10, 0)
```

# ValueError

```python
def add(x, y):
    return int(x) + int(y)

try:
    result = add(10, "5")
    print("Result:", result)
except ValueError:
    print("Invalid value: cannot convert to integer")
```

```python
try:
    num = int("hello")
except ValueError:
    print("Invalid value: cannot convert to integer")
```

Finally, if try and except operate successfully.

```python
try:
    result = 10 // 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("This block always executes")
```

IndexError: out of index or what

```python
try:
    nums = [1, 2, 3]
    print(nums[5])

except IndexError:
    print(f"Invalid index access. Index should be less than {len(nums)}")
```

```python
try:
    import some_module_that_does_not_exist
except ImportError:
    print("Module could not be imported")
else:
    print("Import succeeded!")
    some_module_that_does_not_exist.greet()
    print(some_module_that_does_not_exist.value)
```

# Raise: Allow in python, but not allow in your real usage

```python
try:
    age = -5
    if age < 0:
        raise ValueError("Age cannot be negative")
except ValueError as e:
    print(f"Manually raised: {e}")
```
